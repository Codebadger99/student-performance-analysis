```python
nan_cols = ['Teacher_Quality', 'Parental_Education_Level', 'Distance_from_Home']

for cols in nan_cols:
    df[cols] = df[cols].fillna(df[cols].mode()[0])
```

```python
df.duplicated().sum()
```

```python
plt.figure(figsize=(15, 5))

sns.boxplot(df)

plt.show()
```

```python
outlier_cols = ['Hours_Studied', 'Tutoring_Sessions', 'Exam_Score']

for cols in outlier_cols:
    Q3 = df[cols].quantile(0.25)
    Q1 = df[cols].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q3 - 1.5 * IQR
    higher = Q3 + 1.5 * IQR

    df[cols] = df[cols].clip(lower, higher)   
```

```python
plt.figure(figsize=(15, 5))

sns.boxplot(df)

plt.title("After clipping outliers")

plt.show()
```

```python
tier_map = {"Low": 1, "Medium": 2, "High": 3}
internet_map = {"No": 1, "Yes": 3}

def map_tutoring(sessions):
  if sessions == 0:
    return 1
  elif sessions <= 2:
    return 2
  else:
    return 3

df["PI_mapped"] = df["Parental_Involvement"].map(tier_map)
df["AR_mapped"] = df["Access_to_Resources"].map(tier_map)
df["IA_mapped"] = df["Internet_Access"].map(internet_map)
df["Tut_mapped"] = df["Tutoring_Sessions"].apply(map_tutoring)

df["Support_System_Score"] = (
    df["PI_mapped"] + df["AR_mapped"] + df["IA_mapped"] + df["Tut_mapped"]
)
  
df["Engagement_score"] = df['Hours_Studied'] * (df['Attendance'] / 100)
```

```python
def map_instensity(hours):
    if hours < 15:
        return "Light Study"
    elif 15 <= hours <= 25:
        return "Moderate"
    else:
        return "Intense"



df['Study_Intensity'] = df['Hours_Studied'].apply(map_instensity)
```

```python
def map_sleep_hours(hours):
    if hours < 6:
        return "Insufficient"
    elif 6 <= hours <= 8:
        return "Recommended"
    else:
        return "High"



df['Sleep_Adequacy'] = df['Sleep_Hours'].apply(map_sleep_hours)
```

```python
df["Exam_Delta"] = df["Exam_Score"] - df["Previous_Scores"]


Q4 = df["Exam_Score"].quantile(0.25)
Q6 = df["Exam_Score"].quantile(0.75)


def classify_performance(score):
  if score <= Q4:
    return "Underperformers"
  elif score <= Q6:
    return "Average"
  else:
    return "Top Performers"


df["Performance_Tier"] = df["Exam_Score"].apply(classify_performance)
```
