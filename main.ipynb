```python
import pandas as pd

import matplotlib.pyplot as plt

import seaborn as sns

import numpy as np

sns.set_style('darkgrid')

sns.set_palette('dark')

pd.set_option('display.max_columns', None)

pd.set_option('display.max_rows', None)
```

```python
df = pd.read_csv('StudentPerformanceFactors.csv')

df.head()
```

```python
df["Exam_Score"] = df['Exam_Score'].clip(upper=100)
```

```python
df.describe()
```

```python
cols = ["Teacher_Quality", "Parental_Education_Level", "Distance_from_Home"]

for col in cols:
    df[col] = df[col].fillna("Unknown")
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
plt.figure(figsize=(14, 5))

sns.boxplot(df)

plt.show()
```

```python
outliers = ["Hours_Studied", "Exam_Score", "Tutoring_Sessions"]

for outlier in outliers:
    Q1 = df[outlier].quantile(0.25)
    Q3 = df[outlier].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - IQR * 1.5
    higher = Q3 + IQR * 1.5

    outlier_rows = df[(df[outlier] < lower) | (df[outlier] > higher)]

    print(outlier_rows)

    print(f"Outliers in {outlier}: {len(outlier_rows)} rows found")
```

```python

df['Study_Commitment_Index'] = df['Hours_Studied'] * (df['Attendance'] / 100)

tier_map = {'Low': 1, 'Medium': 2, 'High': 3, 'Unknown': 2}
internet_map = {'No': 0, 'Yes': 1}

df['Support_System_Index'] = (
    df['Parental_Involvement'].map(tier_map) +
    df['Access_to_Resources'].map(tier_map) +
    df['Internet_Access'].map(internet_map)
)

df['Score_Improvement'] = df['Exam_Score'] - df['Previous_Scores']

def categorize_study_intensity(hours):
    if hours < 15:
        return 'Low (<15 hrs)'
    elif 15 <= hours <= 25:
        return 'Moderate (15-25 hrs)'
    else:
        return 'High (>25 hrs)'

df['Study_Intensity_Group'] = df['Hours_Studied'].apply(categorize_study_intensity)

def sleep_category(hours):
    if hours < 6:
        return 'Sleep Deprived (<6 hrs)'
    elif 6 <= hours <= 8:
        return 'Optimal (6-8 hrs)'
    else:
        return 'High (>8 hrs)'

df['Sleep_Category'] = df['Sleep_Hours'].apply(sleep_category)

ordinal_mapping = {
    'Low': 1, 
    'Medium': 2, 
    'High': 3, 
    'Unknown': 2,
    'Negative': 1, 
    'Neutral': 2, 
    'Positive': 3,
    'High School': 1, 
    'College': 2, 
    'Postgraduate': 3,
    'Near': 1, 
    'Moderate': 2, 
    'Far': 3
}

ordinal_cols = [
    'Parental_Involvement', 
    'Access_to_Resources', 
    'Motivation_Level',
    'Teacher_Quality', 
    'Peer_Influence', 
    'Family_Income', 
    'Parental_Education_Level', 
    'Distance_from_Home'
]

for col in ordinal_cols:
    df[f'{col}_Num'] = df[col].map(ordinal_mapping)
```

```python
plt.figure(figsize=(15, 7))

sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")

plt.show()
```
